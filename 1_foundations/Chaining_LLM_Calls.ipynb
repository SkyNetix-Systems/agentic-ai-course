# Lab 2: Chaining LLM Calls in Python 🔗🤖

## Goal
Learn how to use the answer from one LLM call as the input to a second LLM call.

**Simple example:**
1. Ask the AI to write a short story idea.
2. Send that idea to the AI again and ask it to create a title.

This is called **chaining** because the output of the first call becomes the input of the next call.

## Step 1: Import the libraries

`os` reads environment variables, `load_dotenv()` loads the `.env` file, and `OpenAI` creates an API client.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
````

## Step 2: Load the API key and create the client

Make sure your project folder contains a `.env` file with a valid key:

```text
OPENAI_API_KEY=your_actual_api_key_here
```

Keep your API key private.

````python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=api_key)
````

## Step 3: Make the first LLM call

Ask the AI to create a short story idea. We save its answer in `story_idea`.

````python
response1 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {"role": "user", "content": "Give me a simple story idea about a robot that helps a student."}
    ]
)

story_idea = response1.choices[0].message.content
print("Story idea:")
print(story_idea)
````

## Step 4: Make the second LLM call using the first answer

Notice that we insert `story_idea` into the second prompt. This connects the two calls.

````python
response2 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {
            "role": "user",
            "content": f"Create a catchy title for this story idea:\n\n{story_idea}"
        }
    ]
)

story_title = response2.choices[0].message.content
print("\nStory title:")
print(story_title)
````

## Step 5: See the chain

The flow is:

**First LLM call:** Create a story idea  
⬇️ `story_idea`  
**Second LLM call:** Create a title from that idea

The key line is:

```python
f"Create a catchy title for this story idea:\n\n{story_idea}"
```

`{story_idea}` inserts the first call's answer into the second call's prompt.

## Quick recap
- An LLM call sends a prompt to a model.
- The first call returns a story idea.
- The second call uses that idea as its input.
- This sequence is called **LLM chaining**.
