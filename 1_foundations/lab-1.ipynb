```python
# Import required libraries.
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

# Load environment variables from the .env file.
load_dotenv()

# Read the OpenAI API key from environment variables.
openai_api_key = os.getenv("OPENAI_API_KEY")

# Create an OpenAI client using the API key.
client = OpenAI(api_key=openai_api_key)

# Define the user's prompt.
request = "Please tell me a funny joke."

# Create messages containing system instructions and the user prompt.
messages = [
    {"role": "system", "content": "You are a funny assistant. Reply in about 10 lines."},
    {"role": "user", "content": "Tell me a funny joke."}
]

# Send the messages to the AI model.
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages
)

# Extract the AI-generated text.
content = response.choices[0].message.content

# Format the response as Markdown.
markdown_output = Markdown(content)

# Display the formatted response in Jupyter Notebook.
display(markdown_output)
```
