```python
# Import libraries
import os
from dotenv import load_dotenv
from openai import OpenAI

# Load the API key from the .env file
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

# Create the OpenAI client
client = OpenAI(api_key=api_key)

# -----------------------------
# LLM CALL 1: Create a story idea
# -----------------------------
response1 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {
            "role": "user",
            "content": "Give me a simple story idea about a robot that helps a student."
        }
    ]
)

# Save the first AI answer
story_idea = response1.choices[0].message.content
print("STORY IDEA:")
print(story_idea)

# -----------------------------
# LLM CALL 2: Create a title using the story idea
# -----------------------------
response2 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {
            "role": "user",
            "content": f"Create a catchy title for this story idea:\n\n{story_idea}"
        }
    ]
)

# Save and print the second AI answer
story_title = response2.choices[0].message.content
print("\nSTORY TITLE:")
print(story_title)
```
