# Lab 1: Your First OpenAI API Call 🤖

In this lab, you will send a simple prompt to an AI model and display its answer.

**Before you start:** Put your valid OpenAI API key in the `.env` file in your project folder:

```text
OPENAI_API_KEY=your_actual_api_key_here
```

Keep your API key private. Do not share it or upload it to GitHub.

## Step 1: Import the tools

- `os` lets Python read environment variables.
- `load_dotenv()` loads values from your `.env` file.
- `OpenAI` creates a client for calling the OpenAI API.
- `display()` shows the answer nicely in the notebook.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Markdown
````

## Step 2: Load your API key

`os.getenv()` reads the value of `OPENAI_API_KEY` from the environment.

````python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
````

## Step 3: Connect to OpenAI

`OpenAI()` creates the client that will send requests to the API.

````python
client = OpenAI(api_key=api_key)
````

## Step 4: Give the AI instructions and a question

- `system` tells the AI how to respond.
- `user` contains the question or task.

````python
messages = [
    {"role": "system", "content": "You are a funny assistant. Reply in about 10 lines."},
    {"role": "user", "content": "Tell me a funny joke."}
]
````

## Step 5: Send the messages to the AI

`create()` sends the messages to the selected model and returns its response.

````python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)
````

## Step 6: Display the answer

`response.choices[0].message.content` gets the generated text. `display(Markdown(...))` shows it as formatted Markdown.

````python
answer = response.choices[0].message.content
display(Markdown(answer))
````

## Quick recap 🎓

1. Load the API key.
2. Create the OpenAI client.
3. Write system and user messages.
4. Send the request.
5. Display the AI's answer.

**If you see an API-key or model error:** check that your `.env` file contains a valid key and that the model is available to your API account.
